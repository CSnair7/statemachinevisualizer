```python
import graphviz
dot = graphviz.Digraph('G', filename="Graph.png")

dot.attr('node', shape='square')
# Define all data in one place
nodes = [('A', 'Alpha'), ('B', 'Beta'), ('C', 'Gamma')]
edges = [('A', 'B'), ('B', 'C'), ('A', 'C')]

# Declare nodes in a brief loop
for node_id, label in nodes:
    dot.node(node_id, label)

# Mass declare edges with a single method call
dot.edges(edges)
dot.render(filename='process_graph', directory='output_directory', view=True)
```

```python
import graphviz

# 1. Create a directed graph and specify the output format
dot = graphviz.Digraph(comment='My Flowchart', format='png')

# 2. Add nodes and edges
dot.node('A', 'Start')
dot.node('B', 'Process')
dot.node('C', 'End')

dot.edge('A', 'B')
dot.edge('B', 'C')

# 3. Render and save the image
# This creates 'process_graph.gv' (source) and 'process_graph.gv.png' (image)
dot.render(filename='process_graph', directory='output_directory', view=True)
```
