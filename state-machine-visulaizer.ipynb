```python
import graphviz
```

```python
dot = graphviz.Digraph('G')

# Define all data in one place
nodes = [('A', 'Alpha'), ('B', 'Beta'), ('C', 'Gamma')]
edges = [('A', 'B'), ('B', 'C'), ('A', 'C')]


dot.attr('node', shape='doublecircle')


# Declare nodes in a brief loop
for node_id, label in nodes:
    dot.node(node_id, label)

# Mass declare edges with a single method call
dot.edges(edges)
dot.edge('A', 'A', label='2x')
dot.view()
```

```python
# Instantiate graph with graph name and file name
g = graphviz.Digraph('G', filename='cluster.gv')

# Name cluster with lowercase 'cluster' so its recognized as subgraph
with g.subgraph(name='cluster_0') as c:
    # graph attr:
    c.attr(style='filled', color='lightgrey', shape='circle', label='Group A')

    #node attr
    c.node_attr.update(style='filled', color='white')
    c.edges([('a0', 'a1'), ('a1', 'a2'), ('a2', 'a3')])

with g.subgraph(name='cluster_1') as b:
    b.attr(color='red')
    b.node_attr.update(style= 'filled', color='yellow')
    b.edges([('b0', 'b1'), ('b1', 'b2'), ('b2', 'b3')])
    b.attr(label='process #2')

g.edge('start', 'a0')
g.edge('start', 'b0')
g.edge('a1', 'b3')
g.edge('b2', 'a3')
g.edge('a3', 'a0')
g.edge('a3', 'end')
g.edge('b3', 'end')

g.node('start', shape='diamond')
g.node('end', shape='Msquare')

g.view()
```
